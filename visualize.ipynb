```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud

df = pd.read_csv("cleaned_data.csv")

print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["Experience (Years)"],
    bins=10,
    kde=True
)

plt.title("Experience Distribution")
plt.xlabel("Years of Experience")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["Projects Count"],
    bins=10,
    kde=True
)

plt.title("Projects Distribution")
plt.show()
```

```python
plt.figure(figsize=(5,4))

sns.countplot(
    x="Recruiter Decision",
    data=df
)

plt.title("Hiring Distribution")
plt.show()
```

```python
sns.boxplot(
    x="Recruiter Decision",
    y="Experience (Years)",
    data=df
)

plt.title("Experience vs Hiring")
plt.show()
```

```python
sns.boxplot(
    x="Recruiter Decision",
    y="Projects Count",
    data=df
)

plt.title("Projects vs Hiring")
plt.show()
```

```python
sns.regplot(
    data=df,
    x="Experience (Years)",
    y="Recruiter Decision",
    logistic=True
)

plt.title("Experience vs Hiring Probability")
plt.show()
```

```python
sns.pairplot(
    df[
        [
            "Experience (Years)",
            "Projects Count",
            "Recruiter Decision"
        ]
    ],
    hue="Recruiter Decision"
)

plt.show()
```

```python
text = " ".join(df["Skills"])

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(text)

plt.imshow(wordcloud)
plt.axis("off")
plt.title("Most Common Skills")
plt.show()
```
